```python

%sql
use instagram.goldlayer;
CREATE OR REPLACE VIEW instagram.goldlayer.vw_social_lifestyle_impact AS
SELECT 
    e.daily_active_minutes_instagram, 
    (e.time_on_reels_per_day + 
     e.time_on_explore_per_day + 
     e.time_on_messages_per_day + 
     e.time_on_feed_per_day) AS total_time_consumed,
    s.*
FROM instagram.goldlayer.fact_user_engagement e
JOIN instagram.goldlayer.dim_social s ON e.user_id = s.user_id;
```
