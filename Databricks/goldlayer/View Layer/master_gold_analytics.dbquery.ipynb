```sql
USE instagram.goldlayer;

CREATE OR REPLACE VIEW vw_master_analytics_gold AS
SELECT 
    *,
    (time_on_reels_per_day + 
     time_on_explore_per_day + 
     time_on_messages_per_day + 
     time_on_feed_per_day) AS total_time_consumed,
    
    CASE 
        WHEN user_engagement_score > 80 THEN 'High'
        WHEN user_engagement_score > 50 THEN 'Medium'
        ELSE 'Low' 
    END AS engagement_category
FROM instagram.silverlayer.silver_instagram_users;
```
