```python
df_profiles=spark.sql("select * from flight_delay.default.profiles")
df_profiles.show()
```

```python
df_profiles = df_profiles \
    .withColumn("city", split(df_profiles["location"], ",").getItem(0)) \
    .withColumn("state", split(df_profiles["location"], ",").getItem(1))

df_bio_data=df_profiles.select(
    df_profiles["name"].alias("Name"),
    df_profiles["email"].alias("Email"),
    df_profiles["city"].alias("City"),
    df_profiles["current_role"].alias("Current Role"),
    df_profiles["current_company"].alias("Current Company"),
)
```
