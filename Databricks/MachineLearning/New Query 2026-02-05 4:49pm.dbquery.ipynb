```sql
USE instagram.model_output;
CREATE OR REPLACE TABLE fact_model_insights AS
SELECT 

    m.*,

    s.pred_stress_category,

    e.predicted_engagement_score,

    seg.user_segment_id

FROM vw_executive_summary m

JOIN ml_stress s ON m.user_id = s.user_id

JOIN ml_engagement e ON m.user_id = e.user_id

JOIN ml_segments seg ON m.user_id = seg.user_id;



```
