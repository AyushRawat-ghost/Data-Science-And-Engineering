```python
%sql
CREATE OR REPLACE TABLE instagram.model_output.model_combined_output AS
SELECT 
    m.*,
    s.pred_stress_category as stress_predicted,
    e.prediction as engagement_predicted,
    seg.user_segment_id
FROM instagram.goldlayer.vw_executive_engagement m
JOIN instagram.model_output.ml_stress s 
    ON m.user_id = s.user_id
JOIN instagram.model_output.ml_engagement e 
    ON m.user_id = e.user_id
JOIN instagram.model_output.ml_segments seg 
    ON m.user_id = seg.user_id;
```
